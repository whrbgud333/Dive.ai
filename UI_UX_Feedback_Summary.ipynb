# AI Chat LGM UI/UX 개선 제안 및 기능 구현 정리

본 문서는 사용자 피드백을 바탕으로 한 AI Chat 프로젝트의 UI/UX 개선 사항과 신규 기능 아이디어를 정리한 리포트입니다.

## 1. 화면 1: 콘텐츠 정책 및 가이드라인
### [피드백]
- 호러물 출력 제한 문구가 호러 장르 선호 유저의 유입을 저해할 가능성이 있음.

### [개선 방향]
- 경고성 문구보다는 장르적 허용 범위나 AI의 안전 가이드라인을 완곡하게 표현.
- 특정 모드(세이프 모드 vs 다크 모드) 선택권을 주어 유저의 기대치를 조절하는 방안 검토.

## 2. 화면 2: 인터렉션 및 캐릭터 제어 (섹션 A, B, C)
### [피드백]
- **섹션A (토글):** 자동 진행 버튼 클릭 시 진행되는 턴을 1~3턴 중 선택할 수 있도록 옵션 보충.
- **섹션B (입력):** 상태창을 독립시키거나(On/Off), 캐릭터 등장 전용 명령어/버튼을 추가하여 유저의 스토리 주도권 강화.
- **섹션C (페르소나):** 캐릭터 설정에 '빙의'할 수 있는 On/Off 버튼 추가 (AI 인식률 테스트 필요).

### [구현 아이디어]
- `Auto-play Turns`: 드롭다운 메뉴를 통해 자동 진행 횟수 설정.
- `Character Summoner`: 특정 버튼 클릭 시 시스템 프롬프트를 통해 캐릭터를 상황에 즉시 투입.

## 3. 화면 3: 생성 프로세스 시각화
### [피드백]
- AI 생성 중 '잠시 기다려주세요' 등의 로딩 문구 표시.
- 토큰 소모 과정이나 생성 진행도를 시각화하여 로딩 중에도 정보를 제공.

### [개선 방향]
- 실시간 텍스트 렌더링과 함께 'AI가 시나리오를 구상 중입니다...'와 같은 상태 메시지 노출.

## 4. 화면 4: 시나리오 제어 및 진행도 표시
### [피드백]
- **사용자 오리지널 버튼:** 정해진 스토리가 아닌 사용자 의도대로 시나리오를 이끄는 On/Off 기능.
- **기승전결 인디케이터:** 현재 스토리의 진행 단계(기-승-전-결)를 시각화하여 한눈에 파악.
- **상태창 탭 확장:** 상태창 내에 별도 탭을 만들어 추가 정보를 관리.

### [구현 아이디어]
- 진행도 바(Progress Bar)를 상단 또는 상태창에 배치하여 서사적 위치 확인.

## 5. 화면 5: 기억 시스템 및 페르소나 생성 최적화
### [피드백]
- **AI 생성 페르소나 세부화:** AI가 생성한 페르소나 세부 사항이 스토리 기억(유저 노트)에 연동되도록 구성.
- **입력 UX 간소화:** 성격, 직업, 배경, 외형 등을 각각 입력하지 않고, 페르소나 생성 단계에서 통합 관리하여 유저 피로도 감소.

### [개선 방향]
- `Persona Creation Wizard`: 통합 입력 폼을 제공하여 한 번에 페르소나를 완성.
- `Token Efficiency`: AI 생성 세부 사항이 늘어날 경우를 대비해 핵심 키워드 위주로 요약하여 컨텍스트에 주입.
